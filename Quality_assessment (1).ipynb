```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------
# Table 5.1 data
# -----------------------------
df = pd.DataFrame({
    "Sample": [
        "SRR26205319", "SRR26205320", "SRR26205321", "SRR26205322",
        "SRR18324548", "SRR18324549", "SRR18324550", "SRR18324551"
    ],
    "Condition": [
        "AD", "AD", "AD", "AD",
        "MB", "MB", "MB", "MB"
    ],
    "Total Reads": [
        11927701, 39125509, 36832051, 12681624,
        2000010, 2000016, 1000004, 2000012
    ],
    "Mapped Reads": [
        11822301, 38615881, 36336340, 12386401,
        1980846, 1980740, 987897, 1975667
    ],
    "Alignment Rate (%)": [
        99.12, 98.70, 98.65, 97.67,
        99.04, 99.04, 98.79, 98.78
    ],
    "GC Content (%)": [
        45, 45, 49, 46,
        47, 47, 49.5, 49.5
    ],
    "Read Length (bp)": [
        50, 50, 51, 51,
        49, 49, 49, 49
    ],
    "Total Peaks": [
        11689, 21475, 39422, 12970,
        236, 235, 1987, 8797
    ]
})

# -----------------------------
# Parameters to plot
# -----------------------------
parameters = [
    "Total Reads",
    "Mapped Reads",
    "Alignment Rate (%)",
    "GC Content (%)",
    "Read Length (bp)",
    "Total Peaks"
]

# Normalize each parameter to 0–100
plot_df = df.copy()

for col in parameters:
    plot_df[col] = (
        (plot_df[col] - plot_df[col].min()) /
        (plot_df[col].max() - plot_df[col].min()) * 100
    )

# Convert to long format
long_df = plot_df.melt(
    id_vars=["Sample", "Condition"],
    value_vars=parameters,
    var_name="Parameter",
    value_name="Normalized Value"
)

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(18, 8))

sns.barplot(
    data=long_df,
    x="Sample",
    y="Normalized Value",
    hue="Parameter",
    palette="Set2"
)

plt.title(
    "Sequencing Quality Control and Peak-Calling Summary",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Sample", fontsize=13)
plt.ylabel("Normalized Value (0–100)", fontsize=13)

plt.xticks(rotation=45, ha="right")
plt.ylim(0, 110)

plt.legend(
    title="Parameter",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ============================================================
# Table 5.1 data
# ============================================================

df = pd.DataFrame({
    "Sample": [
        "SRR26205319", "SRR26205320", "SRR26205321", "SRR26205322",
        "SRR18324548", "SRR18324549", "SRR18324550", "SRR18324551"
    ],

    "Condition": [
        "AD", "AD", "AD", "AD",
        "MB", "MB", "MB", "MB"
    ],

    "Total Reads": [
        11927701, 39125509, 36832051, 12681624,
        2000010, 2000016, 1000004, 2000012
    ],

    "Mapped Reads": [
        11822301, 38615881, 36336340, 12386401,
        1980846, 1980740, 987897, 1975667
    ],

    "Alignment Rate (%)": [
        99.12, 98.70, 98.65, 97.67,
        99.04, 99.04, 98.79, 98.78
    ],

    "GC Content (%)": [
        45.0, 45.0, 49.0, 46.0,
        47.0, 47.0, 49.5, 49.5
    ],

    "Read Length (bp)": [
        50, 50, 51, 51,
        49, 49, 49, 49
    ],

    "Total Peaks": [
        11689, 21475, 39422, 12970,
        236, 235, 1987, 8797
    ]
})


# ============================================================
# Publication-style settings
# ============================================================

sns.set_theme(style="white", context="paper")

# Color-blind-friendly colors
colors = {
    "AD": "#0072B2",   # blue
    "MB": "#D55E00"    # orange
}


# ============================================================
# Create 2 × 3 figure
# ============================================================

fig, axes = plt.subplots(
    2, 3,
    figsize=(14, 8),
    dpi=300
)

# ------------------------------------------------------------
# Panel A: Total reads
# ------------------------------------------------------------

sns.barplot(
    data=df,
    x="Sample",
    y="Total Reads",
    hue="Condition",
    palette=colors,
    legend=False,
    ax=axes[0, 0]
)

axes[0, 0].set_title("A. Total reads", fontweight="bold")
axes[0, 0].set_xlabel("")
axes[0, 0].set_ylabel("Reads")
axes[0, 0].tick_params(axis="x", rotation=45)


# ------------------------------------------------------------
# Panel B: Mapped reads
# ------------------------------------------------------------

sns.barplot(
    data=df,
    x="Sample",
    y="Mapped Reads",
    hue="Condition",
    palette=colors,
    legend=False,
    ax=axes[0, 1]
)

axes[0, 1].set_title("B. Mapped reads", fontweight="bold")
axes[0, 1].set_xlabel("")
axes[0, 1].set_ylabel("Mapped reads")
axes[0, 1].tick_params(axis="x", rotation=45)


# ------------------------------------------------------------
# Panel C: Alignment rate
# ------------------------------------------------------------

sns.barplot(
    data=df,
    x="Sample",
    y="Alignment Rate (%)",
    hue="Condition",
    palette=colors,
    legend=False,
    ax=axes[0, 2]
)

axes[0, 2].set_title("C. Alignment rate", fontweight="bold")
axes[0, 2].set_xlabel("")
axes[0, 2].set_ylabel("Alignment rate (%)")
axes[0, 2].set_ylim(95, 100)
axes[0, 2].tick_params(axis="x", rotation=45)


# ------------------------------------------------------------
# Panel D: GC content
# ------------------------------------------------------------

sns.barplot(
    data=df,
    x="Sample",
    y="GC Content (%)",
    hue="Condition",
    palette=colors,
    legend=False,
    ax=axes[1, 0]
)

axes[1, 0].set_title("D. GC content", fontweight="bold")
axes[1, 0].set_xlabel("")
axes[1, 0].set_ylabel("GC content (%)")
axes[1, 0].set_ylim(40, 55)
axes[1, 0].tick_params(axis="x", rotation=45)


# ------------------------------------------------------------
# Panel E: Read length
# ------------------------------------------------------------

sns.barplot(
    data=df,
    x="Sample",
    y="Read Length (bp)",
    hue="Condition",
    palette=colors,
    legend=False,
    ax=axes[1, 1]
)

axes[1, 1].set_title("E. Read length", fontweight="bold")
axes[1, 1].set_xlabel("")
axes[1, 1].set_ylabel("Read length (bp)")
axes[1, 1].set_ylim(45, 55)
axes[1, 1].tick_params(axis="x", rotation=45)


# ------------------------------------------------------------
# Panel F: Total peaks
# ------------------------------------------------------------

sns.barplot(
    data=df,
    x="Sample",
    y="Total Peaks",
    hue="Condition",
    palette=colors,
    legend=False,
    ax=axes[1, 2]
)

axes[1, 2].set_title("F. Total peaks", fontweight="bold")
axes[1, 2].set_xlabel("")
axes[1, 2].set_ylabel("Number of peaks")
axes[1, 2].tick_params(axis="x", rotation=45)


# ============================================================
# General formatting
# ============================================================

for ax in axes.flat:

    # Remove top and right borders
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    # Light horizontal grid
    ax.grid(
        axis="y",
        linestyle="--",
        linewidth=0.5,
        alpha=0.4
    )

    ax.set_axisbelow(True)

    # Consistent font
    ax.tick_params(
        axis="both",
        labelsize=9
    )


# ============================================================
# Single legend for the entire figure
# ============================================================

from matplotlib.patches import Patch

legend_elements = [
    Patch(facecolor=colors["AD"], label="AD"),
    Patch(facecolor=colors["MB"], label="MB")
]

fig.legend(
    handles=legend_elements,
    loc="upper center",
    ncol=2,
    frameon=False,
    bbox_to_anchor=(0.5, 1.02),
    fontsize=10
)


# ============================================================
# Figure title
# ============================================================

fig.suptitle(
    "Sequencing quality control and peak-calling summary",
    fontsize=15,
    fontweight="bold",
    y=1.06
)


# ============================================================
# Layout
# ============================================================

plt.tight_layout()

# Save high-resolution figure
plt.savefig(
    "Figure_5_1_Sequencing_QC_Peak_Summary.png",
    dpi=600,
    bbox_inches="tight"
)

plt.savefig(
    "Figure_5_1_Sequencing_QC_Peak_Summary.pdf",
    bbox_inches="tight"
)

plt.show()
```
